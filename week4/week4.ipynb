## Loops

Let's start with a simple list that has a sequence of "actions"

```python
routine = ["wake up","eat breakfast","drink coffee","brush teeth","put on shoes","go to work"]
```

```python
print(routine[0])
```

Below you will see a `for` loop that will iterate through every item in the list. Notice the syntax: the indented portion is the portion of the for loop that is repeated

```python
print(routine)

for step in routine: # extract each element in routine one by one, referring to it as "step"
    print(step)
```

A `for` loop consists of several elements:
- the `for` keyword, which inidicates we are declaring a looping structure
- an incremented variable (step), which will represent each element in the list. it can be called whatever you want. often it is called "i" as a generic term, but it often makes sense to refer to whatever a specific entry is within your list.
- the `in` keyword, followed by a list, collection, or sequence (anything that can be iterated through)
- the **:** (colon) character, which tells python that the looping lines of code are to follow. The lines below a colon will be indented one level further than we currently are.
- and finally, the functions to loop. these are functions that are repeated for every element in the list

```python
fruit = "banana"

for i in fruit:
    print(i.upper())
```

```python
grades = [56,59,1,99,55,80,90]
sum_of_grades = 0  # initializing the sum variable at zero...

for grade in grades:
    sum_of_grades = sum_of_grades + grade # .. so i can add to it 
    # with each step of the loop

print(sum_of_grades)
print(sum_of_grades/len(grades))
```

## Function declarations!

We can create custom functions that do specific tasks in Python! I'll use the above running time in seconds conversion as an example!

```python
# this function takes a string formatted like "30:15"
# and returns an int representing total time in seconds

def calculate_seconds(time_string): # def + function name + arguments in parentheses + colon
    # we are expecting a string formatted like "30:15" as input
    # which will represented by the placeholder "time_string"
    minutes,seconds = time_string.split(":") # splitting the string
    total_seconds = int(minutes)*60 + int(seconds) # converting + calculating total seconds
    return total_seconds

print(calculate_seconds("40:45"))
```

Function declarations are made of the following elements:
- the `def` keyword: tells python that we are defining a new function!
- the name of the new function(`calculate_seconds`)
- the arguments: multiple arguments are separated by commas
- the arguments are then processed by the lines of code that are indented below the "def" line
- `return` specifies what data is output by the function

```python
# another example, for generating angry text from normal text

def angry(text):
    text = text.upper() # makes text uppercase
    exclamations = "!"*len(text)
    return text + exclamations

print(angry("hello world"))
```

## Conditionals

Conditionals, like `for` loops, are another fundamental structure found across most programming languages

Conditionals rely on comparison operators, which compare two values and return a value of `True` or `False`

- `x == y` : Are the values equal? True for yes, False for no
- `x != y` : Are the values unequal?
- `x < y` : Is the first value less than the second value?
- `x > y` : Is the first value greater than the second value?
- `x <= y` : Is the first value less than **or** equal to the second value?
- `x >= y` : Is the first value greater than **or** equal to the second value?

Comparison operators can also be combined with **logical operators** which allow you to check more than one thing at once (or, in the case of the `not` operator, the opposite of a True/False value)

- `x > y and a == b` : are both conditions true?
- `x < y or a > b` : is at least one of the conditions true?

```python
x = 5
y = 6
a = 1
b = 1

print(x>y)
print(x<y)
print(x==y)
print(x<y and a!=b)
print(x>y or a==b)
```

In Python we can run a block of code *conditionally* – that is, only run the code if a condition is true – using the `if` and `else` syntax:

```python
temperature = 40

if temperature < 60:
    print("put on a jacket")
else:  #else catches all data that doesn't match the condition
    print("no jacket needed!")
```

In the above simple example we see the following elements:
- the `if` keyword, which initializes the conditional statement
- the comparison (temperature < 60), which is evaluated as True or False
- if the comparison evaluates to `True`, the indented portion directly below the if statement runs
- if the comparison is not `True`, the indented portion below `else:` runs

We can also stack conditions using the else if, or `elif` keyword:

temperature = 30

if temperature >= 80:
    print("wear shorts")
elif temperature >= 65:
    print("wear pants")
elif temperature >= 50:
    print("wear a sweater")
elif temperature >=35:
    print("wear a jacket")
else:
    print("stay home")

We can add logical operators (like `and` & `or`) to combine comparisons in conditionals:

```python
temperature = 70

if temperature >= 65 and temperature <= 75:
    print("the weather is comfortable")
elif temperature < 32 or temperature > 95:
    print("the weather is extreme")
```

### Using conditionals to traverse and filter a dataset

We can combine conditionals and `for` loops to filter a dataset, looking for certain matching criteria within all the elements in a list.

In the below example, we will go through every Beatles album in the set, go through every track in the track list key, and search for the word "love". The output will be a new list of song titles containing the word "love".

```python
for album in beatles_albums:  # this is a nested for loop, in which we iterate through a parent list and its child list
    for track in album["tracks"]:
        print(track)
```

```python
# we want to search for the word "love" in each song title

for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if "love" in track.lower(): # returns true if "love" is found in the string
            # track.lower() converts the text to lower case
            print(track)
```

```python
# I want to catch "loving" and other variants

for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if "Lov" in track: # returns true if "love" is found in the string
            # track.lower() converts the text to lower case
            print(track)
```

```python
# and here's the most robust solution, which could catch lower case variants and
# will IGNORE fragments in words like "clover"

for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if " lov" in (" "+track).lower(): # returns true if "love" is found in the string
            # track.lower() converts the text to lower case
            print(track)
```

```python
# in this example, i'm making a NEW list of only the songs that have "love" in them

love_songs = []  # this is an empty list!

for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # and for each track in each album...
        if " lov" in (" "+track).lower(): # returns true if "love" is found in the string
            # track.lower() converts the text to lower case
            love_songs.append(track)

print(love_songs)
print(len(love_songs))
```

# Working with CSV data

CSV is one of the most common dataset formats, it's basically a spreadsheet with rows and columns
Each row is a new line, each column is separated by a comma
CSVs usually have a "header" row that is in the first row of the document
Each subsequent row is the actual data!

Let's have a look at the MoMA dataset that I've had everyone download (moma_artists.csv)

To work with CSV data in python we need to import the CSV library, by typing `import csv` before attempting to use any csv functions

We can open a file using the following syntax:

`with open(file_name, options) as file:`

```python
import csv

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)  # initializing an object that interprets CSV data
    # the below line will extract all the headings!
    header = next(reader)  # tells the reader to read the first line and move on to the next
    rows = list(reader)    # takes every other row in the dataset and makes it into a list

print(header)
print(rows[0]) # first element in list
```

We can also generate a list of dictionaries that use the header to establish a set of shared keys, assigning values for each row based on those keys/headers.

We can do that using the csv method `DictReader(file)`

```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)  # this will interpret the csv as a list of dictionaries
    data = list(reader)   # store the CSV data as a list of dictionaries we can access later

print(data[-1])
```

```python
# filter by nationality: only Swedish artists

swedish_artists = []   # create an empty list before i run the for loop

for artist in data:  # for each artist in my data...
    if artist["Nationality"] == "Swedish":
        swedish_artists.append(artist)

print(len(swedish_artists))

for artist in swedish_artists[:10]:
    print(artist["DisplayName"])
```

```python
young_artists = []

for artist in data:
    if int(artist["BeginDate"]) > 1990 and artist["Gender"] != "":
        young_artists.append(artist)

print(len(young_artists))

for artist in young_artists[:10]:
    print(artist["DisplayName"])
```

```python
import csv   # brings in all the csv-related functions

# the open function takes these three arguments:
# - the file name / path
# - "r" for read
# - and the encoding type

with open("moma_art.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # read the file as dictionary object
    data = list(reader) # convert that dictionary object into a usable list called "data"

print(data[0]) # prints the first dictionary item
```

```python
ono_art = []

for artwork in data:
    if artwork['Artist'] == "Yoko Ono":
        ono_art.append(artwork)

print(len(ono_art))
```

```python
with open("moma_ono.csv", "w", encoding="utf-8") as file:
    writer = csv.DictWriter(file, ono_art[0].keys()) # new csv from dict, using all the keys from ono_art
    writer.writeheader() # writes a header row for the csv
    writer.writerows(ono_art)
```

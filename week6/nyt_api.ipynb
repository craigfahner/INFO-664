NYT API

```python
import requests
from getpass import getpass
from time import sleep

key = getpass("Paste your NYT API key: ")
```

```python
query = "Carney"

url = f"https://api.nytimes.com/svc/search/v2/articlesearch.json?q={query}&api-key={key}"
response = requests.get(url)

print(response)
```

```python
parsed_nyt_response = response.json()
print(parsed_nyt_response.keys())
```

```python
print(parsed_nyt_response['response'].keys())
```

```python
print(parsed_nyt_response['response']['metadata'])
```

```python
print(parsed_nyt_response['response']['docs'])
```

```python
articles = parsed_nyt_response['response']['docs']
print(articles[0].keys())
```

```python
print(len(articles))
```

```python
results = []

for page in range(0,5):
    url = f"https://api.nytimes.com/svc/search/v2/articlesearch.json?q={query}&page={page}&api-key={key}"
    response = requests.get(url)

    if response.ok:
        parsed = response.json()
        articles = parsed["response"]["docs"]
        results.append(articles)
        article_amount = len(articles)
        print(f"grabbed {article_amount} articles from page {page}")
    else:
        print(f"Page {page} failed with error code {response.status_code}")
    print("waiting 12 seconds before requesting more...")
    sleep(12) # wait 12 seconds so we don't overwhelm nyt api

print(len(results))
```

```python
abstracts = []
search_term = "tariff"
mention_count = 0


for result in results:
    for article in result:
        abstracts.append(article['abstract'])

for abstract in abstracts:
    count = abstract.count(search_term)
    if count > 0:
        mention_count = mention_count + 1
        print(abstract)

print(f"There are {mention_count} articles that include the term '{search_term}'")
```

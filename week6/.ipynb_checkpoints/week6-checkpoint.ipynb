## Saving CSVs

Quick example for how to save a CSV from a list of dictionaries

```python
import csv   # brings in all the csv-related functions

with open("moma_art.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)   # read the file as dictionaries
    data = list(reader)             # convert that into a usable list called "data"

print(data[0].keys())
```

```python
# filtering the large dataset into a smaller list of dictionaries with only works by Yoko Ono

yoko_ono_artworks = [] 

for artwork in data:
    if artwork['Artist'] == "Yoko Ono":
        yoko_ono_artworks.append(artwork)

print(len(yoko_ono_artworks))
```

```python
with open("yoko_ono_artworks.csv", "w", encoding="utf-8") as file:
    writer = csv.DictWriter(file, yoko_ono_artworks[0].keys())  # specifying the keys in yoko_ono_artworks as the columns in the CSV
    writer.writeheader()  # create the first row of the CSV, containing the headers
    writer.writerows(yoko_ono_artworks)
```

## Regular Expressions

```python
caption = "William Wegman, Cotto, 1970. Gelatin silver print, 7 7/8 × 7 3/4 in. (20 × 19.7 cm). Whitney Museum of American Art, New York; purchase with funds from the Mrs. Percy Uris Purchase Fund and the Photography Committee 92.14 © William Wegman"
```

Regular Expressions are a markup language used for identifying patterns in strings, in order to do complex searches. They use a bunch of specific special characters which represent different sequences to be extracted.

```python
import re   # importing regular expressions functions/methods

match = re.search(r"([0-9]{4})", caption)

print(match)

year = match.group(0)

print(year)
```

```python
with open("../week5/moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)   # read the file as dictionaries
    artists_data = list(reader)             # convert that into a usable list called "data"

print(artists_data[0])
```

```python
catherines = []

for artist in artists_data:
    if re.search(r"^[CK]ath\w*", artist['DisplayName']):
        catherines.append(artist)

for catherine in catherines:
    print(catherine['DisplayName'])
```

## Reading JSON files

We read JSON files the same way we read CSVs, except they are naturally interpreted as lists of dictionaries

```python
import json   # import JSON methods

with open("unesco_cultural_sites.json", "r", encoding="utf-8") as file:
    sites = json.load(file)

print(len(sites))
print(type(sites))
print(sites[0].keys())
print(sites[0]['name_en'])
```

## APIs!

APIs are used to query open data providers in real time by grabbing data from a server. The query is provided through a URL, and the response is usually formatted as a JSON object. We will be looking at two different APIs today:

- The MET api, which can
  - Return object ids for search terms
  - Return details of a given object based on an object id
- NYTimes API, which can return lists of metadata on articles based on search terms



```python
import requests       # brings in the functions needed to grab data from the web and deal with responses

# the variables below make up the components of the URL we will use to query the MET api

base_url = "https://collectionapi.metmuseum.org" # the base URL is always the same no matter what MET endpoint we are using
search_path = "/public/collection/v1.1/search" # this is the "endpoint" that deals with searching for object IDs in the collection
search_query = "?q=Antelope" # these are the parameters of the API query. q means  "search term", and "Antelope" is what we are searching for
date_range = "&dateBegin=1900&dateEnd=2000"

url = f'{base_url}{search_path}{search_query}{date_range}'
search_response = requests.get(url)

print(url)
```

```python
print(type(search_response)) # this shows us that what we get back is called a Response
```

```python
print(dir(search_response)) 
```

```python
print(search_response.ok) # returns false if data was received as expected
```

```python
print(search_response.status_code) # 404 means not found, 200 means OK
```

```python
met_objects = search_response.json() # creating a new list of dictionaries by interpreting the response as JSON

print(type(met_objects))
```

```python
print(met_objects.keys()) # shows us what keys are contained in the JSON data from the API
```

```python
print(met_objects['total']) # shows us what's in the "total" key (total amount of objects with the search criteria)
```

```python

antelope_object_ids = met_objects['objectIDs']
print(len(antelope_object_ids)) # this now contains a list of MET Object IDs representing Antelope-related stuff
```

```python
object_path = "/public/collection/v1/objects/" # this is the path of the objects endpoint in the MET API
antelope_item = antelope_object_ids[0]  # this is the first item in the list of object IDs above

url = f"{base_url}{object_path}{antelope_item}"
print(url)

response_object = requests.get(url)

print(response_object.ok)
```

```python
antelope_object_details = response_object.json() # interpreting the object api response as a JSON object

print(antelope_object_details.keys())
```

```python
antelope_object_details['title']
```

```python
antelope_object_details['measurements']
```

## Below is an experiment where I am trying to pull ALL object details about ALL antelope objects in my objectID list


```python
from time import sleep  # this allows us to use time functions related to waiting between operations

# object_path = "/public/collection/v1/objects/" # this is the path of the objects endpoint in the MET API
# antelope_item = antelope_object_ids[0]  # this is the first item in the list of object IDs above

# url = f"{base_url}{object_path}{antelope_item}"
# print(url)

# response_object = requests.get(url)

# print(response_object.ok)

object_path = "/public/collection/v1/objects/"

all_antelope_details = []

for antelope_id in antelope_object_ids: # for each of the object IDs i pulled from the MET api earlier...
    url = f"{base_url}{object_path}{antelope_id}"
    response_object = requests.get(url) # making the request to the server
    all_antelope_details.append(response_object.json())
    print(f"grabbed details from object id {antelope_id} with status code {response_object.status_code}")
    sleep(0.5) # wait half a second before moving on to the next step in the for loop

```

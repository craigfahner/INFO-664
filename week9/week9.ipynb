## Dropping columns and rows

your dataset might have columns that aren't relevant to your analysis, you can get rid of them using `df.drop('column name', axis=1)`

axis=1 specifies that we want to pull out a COLUMN not a row. axis=0 would remove a row

```python
df.head(2)
```

```python
df = df.drop('Unnamed: 0', axis=1)

df.head(3)
```

```python
# I can get rid of multiple unwanted columns using a for loop!

unwanted_columns = ['primaryImageSmall', 'portfolio', 'artistRole', 'artistPrefix', 'artistSuffix', 'artistAlphaSort', 'subregion', 'locale', 'locus', 'excavation', 'river', 'classification', 'rightsAndReproduction', 'linkResource', 'metadataDate', 'repository']

for column in unwanted_columns:
    df = df.drop(column, axis=1)

df.head(3)
```

## Sorting data

we can sort data in different ways using the `df.value_counts()` and `df.sort_values()` methods

`df.value_counts` will count how many times a given value appears in a column, listing the most common values first

```python
df.columns
```

```python
df.value_counts('title') # this will generate a sorted series that is ordered according to how often values appear in the title column
```

```python
title_counts = df.value_counts('title')[:15]
title_counts
```

```python
df.value_counts('accessionYear')
```

`sort_values` will sort the rows of the table according to a specific column. adding `ascending=False` reverses the order

```python
df.sort_values('accessionYear', ascending=False)[:5]
```

## Filtering data

we can filter dataframes using comparison operators like ==

comparing a column to a value will return a Series of every row in the column, and whether that matches the criteria

```python
df['title'] == 'Woman'
```

```python
woman_mask = df['title'] == 'Woman'  #store the series in a new variable called woman_mask

type(woman_mask)
```

The above series is called a *mask* which identifies the filtered content as rows flagged True or False. We can use that *mask* to query the dataframe, which will return a dataframe that has been filtered according to the mask

```python
woman_objects = df[woman_mask]

woman_objects.info()
```

We can also use `df['title'].str.contains('Woman', na=False)`

we need to include na=False to handle missing valies, otherwise the mask would inherit those incorrectly

```python
courbet = df['artistDisplayName'].str.contains('Picasso', na=False)

df[courbet]  # using the above mask to display the dataframe
```

## graphing or plotting data

pandas can generate graphs and charts using `.plot()` . we can generate a variety of types of graphs, including:

- 'bar'
- 'barh' - horizontal bars
- 'pie'
- 'line'
- 'hist' - histogram
- 'scatter'

bar charts are useful when dealing with the `value_counts` Series for a given column

```python
df.value_counts('department').plot(kind = 'barh')
```

```python
df.value_counts('title').nlargest(20).plot(kind = 'bar') # adding nlargest(20) will only include the biggest 20 items
```

```python
df.value_counts('culture').nlargest(20).plot(kind='pie')
```

```python
df.value_counts('artistDisplayName').nlargest(10).plot(kind = 'barh', ylabel='artist name', title='works with "women" in the title')
```

## sorting data:

.value_counts() counts how many times each value appears in a column:

```python
df.value_counts('title')
```

note that this automatically sorts the column in order of most occurrences to least

```python
df.value_counts('department')
```

try `artistDisplayName`, `medium` as well

df.sort_values() will re-order the rows of a table based on a single column's values. `ascending=False` will change the direction

```python
df.sort_values('accessionYear')[:5]
```

## Filtering

rather than returning True or False, the equivalence operator will generate a list of True / False values for each value in the frame

first we find a matching set with an operator, then we use those results to "mask" the complete dataframe into a smaller dataframe

```python
df.title == 'Woman'
```

we can use this as a "mask" to filter the whole set. first, save this into a variable

```python
woman_mask = df['title'] == 'Woman'

# then we use this as a "mask" by placing it in the square brackets on the existing dataframe

woman_objects = df[woman_mask]
```

```python
print("filtered set:",len(woman_objects))
print("full set:",len(df))
```

### Filtering for partial matches

== only finds exact matches. we can use `.str.contains(term)` to search for substrings within a column's values

we need to add the `na=False` argument, which ensures that missing values are treated as `False` rather than other values that str.contains produces



```python
courbet = df['artistDisplayName'].str.contains('Courbet', na=False)
df[courbet]
```

# plotting data

.plot() method generates graphs. `kind=` specifies what kind of chart:
- `bar`
- `barh`: horizontal bar
- `pie`
- `line`
- `hist`
- `box`
- `scatter`
- `hexbin`

Plotting a series draws one bar for each value in its index – this is an apt task for value_counts()

```python
df.value_counts('department').plot(kind = 'bar') # this might take a second
# try pie and barh too
```

```python
df.value_counts('title').plot(kind='bar') # too busy
```

```python
# nlargest(20) only shows 20 largest values in the series (also [:20]) works

df.value_counts('title').nlargest(20).plot(kind='bar') 
```

```python
# adding axis labels

df.value_counts('artistDisplayName').nlargest(20).plot(kind='barh', ylabel='Artist Name', title='Works titled "Women" at the MET')
```

## All about lists

Lists are an ordered collection of values, which are written within square brackets (`[ ]`), and where each value is separated by a comma. Each value can be accessed using an index number that is related to its position within the list. **The first index number in a list is 0**.

```python
print(birds)
print(leap_years)
print(bands)
```

You access data in a list using its **index number**:

```python
print(birds[0]) # this will print the "first" bird in the list (index number 0)
print(birds[0:3]) # this will print a range of birds (index numbers 0 thru 3
print(len(birds))
```

If we want to access a value within a *list of lists* we must use two index numbers:

```python
print(bands[0])   # this gives me the list contained in the "list of lists"
print(bands[0][-1])  # the -1 index gives me the last entry in a list
print(bands[2][1])   # gives me the index 1  member of band at index 2
```

## Searching for a value in a list

```python
rolling_stones = bands[1]
print(rolling_stones)

'Ronnie Wood' in rolling_stones  # returns "true" if the value is in the list
print(rolling_stones.index('Ronnie Wood'))
```

I can reassign the value of any element in a list by redefining the variable with its respective index number

```python
print(rolling_stones)
rolling_stones[3] = "Brian Jones"
print(rolling_stones)
```

## Adding and removing items in a list

```python
print(birds)
birds.append('Tucan')  # append adds new element to end of list
print(birds)
birds.insert(3, 'Cockatoo')  # insert adds a new element at a specific index number, pushing the rest forward
print(birds)
birds.pop(5)  # pop removes an element from a list based on index number
print(birds)
birds.remove("Woodpecker")  # will remove an element based on its value
print(birds)
```

## Reordering a list using sort()

```python
birds.sort()  # sort will put our list of  birds in alphabetical order!
print(birds)
birds.sort(reverse=True) # the reverse=True argument will reverse the list
print(birds)

# the same goes for numbers, they will be put into ascending or descending numerical order:
temperatures_list = [45, 98, 33, 106, 88, 98, 45]
print(temperatures_list)
sorted_temperatures_list = temperatures_list # i've made a copy of the unsorted list into a new variable
sorted_temperatures_list.sort()
print(sorted_temperatures_list)
```

## Simple statistics on a list: min, max, average

```python
print(temperatures_list)
print(min(temperatures_list))  # gives me the minimum value in a list
print(max(temperatures_list))  # gives me the maximum value
print(sum(temperatures_list)/len(temperatures_list))  # takes the sum of the temperatures and divides it by the length of the list
```

### Pulling data out of a list of dictionaries

```python
print(beatles_albums[0])  # gives me a single dictionary item within the list of dictionaries

print(beatles_albums[0]["album_title"])  #pull out album title from the first dictionary

print(beatles_albums[0]["tracks"])       # pulls track list from first dictionary
print(beatles_albums[0]["tracks"][0])     # pulls out first track from first album
```

### Adding new keys and values to a dictionary

My goal here is to look at the "running_time" value, extract a length in seconds by doing some simple math, and then creating a new key in each respective dictionary to represent the "running_time_in_seconds" as an *int*

```python
print(beatles_albums[0]["running_time"])   # accessing the running_time key in the dictionary
```

```python
minute_string, second_string = beatles_albums[0]["running_time"].split(":") # split the running time into two different strings
print(minute_string)
print(second_string)
```

```python
total_seconds = int(minute_string) * 60 + int(second_string)
# converting minute and second strings to ints, multiplying minutes by 60 to get seconds
# and then adding the remaining seconds
print(total_seconds)
```

```python
beatles_albums[0]["running_time_in_seconds"] = total_seconds
print(beatles_albums[0])
```

## Sorting dictionaries

When we are sorting lists, we can use a simple sort method that puts items in alphabetical or numerical order. This is more difficult when dealing with *dictionaries* which have more diverse nested data. It's often the case that we don't merely want to alphabetize, but to sort based on a specific nested quantity.

```python
print(central_park_birds)
```

```python
# sorted() function takes two arguments: one is the dataset to sort
# the other is the logic by which to sort
# in this case, we use a lambda function to specify which key to compare
# the option third argument specifies whether the order should be in descending (reverse) order
sorted_central_park_birds = sorted(central_park_birds, key=lambda item:item["count"], reverse=True)

print(sorted_central_park_birds)
```

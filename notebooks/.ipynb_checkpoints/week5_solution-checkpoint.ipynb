```python
import csv # importing pythons built in functions that handle csv files

with open ("moma_artists.csv", "r", encoding="utf-8") as file: 
    reader = csv.DictReader(file)
    moma_artists = list(reader)
```

```python
year_counts = [] # where i'm gonna store my year and count dictionaries

# i will be populating year_counts with dictionaries that look like this:
#{ 
#  'Year': 1963,
#  'Count': 2
#}

for artist in moma_artists: # iterating through every artist in the set
    birth_year = int(artist['BeginDate']) # storing the begin date so we can look for it in our year_counts array
    did_we_find_year = False # this keeps track of whether the year was found in year_counts
    for logged_year in year_counts: # now we look at every logged year in year_counts (if any)
        if birth_year == logged_year['Year']: # if the moma artist's year is already logged in our list of year counts...
            did_we_find_year = True # flips to true when a match is found so we don't later make a new dictionary for an existing year
            logged_year['Count'] = logged_year['Count'] + 1 # add 1 to the count
    if did_we_find_year == False and birth_year != 0: # if the BeginDate wasn't already in our year_counts list (the nested for loop did not find any matches)
        new_year = {
            'Year': birth_year,
            'Count': 1
        } # Count is 1, since we are making the first entry here for that particular year in the year_counts list
        year_counts.append(new_year) # add that new dictionary to the list

sorted_year_counts = sorted(year_counts, key=lambda year: year["Count"], reverse=True)

print(sorted_year_counts[:10])
```

## Data types in Python

Values in Python can be a number of different data types, such as **int, float, string, bool**

```python
print(type(5))      # INT
print(type(5.5))    # FLOAT (decimal-point number)
print(type("5"))    # STR (string - text)
print(type(True))   # BOOL (boolean - true or false)
```

## Comparison Operators

These are symbols in Python used to compare data, which evaluate to **True** or **False**

```python
print(5 > 3)  # greater than operator
print(5 == 5) # equal to operator
print(5 != 5) # inequality operator
print(3 <= 3) # less than OR equal to operator
print("hello" == "Hello")
print("apple" < "banana")
```

## Arithmetic Operators

These are math operators, used to perform math operations on data

```python
print(7+2)  # addition
print(7-2)  # subtraction
print(7*2)  # multiplication
print(7/2)  # "true" division (results in float, in this case)
print(7//2) # "floor" division (rounds down the result)
print(7%2)  # modulo (remainder)
print(7**2) # exponent
```

## Variable Declaration

Variables in Python can be *declared* using the **=** symbol. The word on the left side of the equals symbol becomes the name of the variable, and its value is assigned on the right side of the equal sign!

```python
name = "Craig"
age = 99
temperature = 0.5
```

```python
print(name)
```

### Naming conventions with variables

- your variable name should represent what it is holding
- names CANNOT start with a number
- names CANNOT have a hyphen in them
- names CAN have underscores in them
  - place_of_birth is a valid variable name using "snake case"
  - USE_UPPER_CASE to designate a constant variable that will not change after its set
- names CANNOT be special words in the python language like "if" and "for"

## Incrementing and modifying variables

Variables can be given new values! They can also be modified by functions, and by arithmetic operators

```python
print(age)
age = age*2
print(age)
```

```python
age += 1  # shorthand for age = age + 1
print(age)
```

```python
age -= 10  # subtract 10
print(age)
age //= 2  # floor divide by 2
print(age)
age *= 3
print(age)
```

## Converting data types

Data can be converted from one type to another using conversion functions:

```python
pi = 3.14159
print(pi)

pi_as_string = str(pi)  # str function converts to a string!
print(pi_as_string)
print(type(pi_as_string))

pi_as_float_again = float(pi_as_string)
print(pi_as_float_again)
print(type(pi_as_float_again))

print(pi_as_float_again>3)
```

## Combining and concatenating strings

Strings can be combined (or *concatenated*) a number of ways. One is to use the *+* operator!

```python
dog_name = "Fido"
dog_breed = "Labrador"
dog_age = 4

#convert the int (dog_age) to string before concatenating
my_dog_string = "My dog's name is " + dog_name + " and he is " + str(dog_age) 
print(my_dog_string)
```

We can also use [F-strings](https://www.w3schools.com/python/python_string_formatting.asp)

```python
my_dog_string = f"My dog's name is {dog_name}. He is {dog_age}, and he is a {dog_breed}."
print(my_dog_string)
```

## Converting string case

There are a bunch of built in string methods for converting to upper, lower, capitalized, and title case

```python
print(dog_name.upper())
print(dog_name.lower())
print(my_dog_string.title())
print(my_dog_string.capitalize())
```

## Searching, counting, extracting

There are some  methods in Python that deal with searching for substrings, etc. One of those is the keyword **in**. 

```python
print("dog" in my_dog_string)  # True if the word "dog" is in the string
```

```python
my_dog_string = my_dog_string + " I really love dogs and dogs are my best friends. I also like cats."
print(my_dog_string)
```

```python
dog_count = my_dog_string.count("dog")
print(f"The word 'dog' appears {dog_count} times.")
```

```python
cat_count = my_dog_string.count("cat")
print(f"The word 'cat' appears {cat_count} times.")
```

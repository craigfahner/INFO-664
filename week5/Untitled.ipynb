```python
import csv   # brings in all the csv-related functions

# the open function takes these three arguments:
# - the file name / path
# - "r" for read
# - and the encoding type

with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # read the file as dictionary object
    data = list(reader) # convert that dictionary object into a usable list called "data"
    
```

```python
# STEP 1: REMOVE ALL ENTRIES WITH NO YEAR!

artists_with_year = [] # new list to hold our artists that have year data

for artist in data: # for each artist in the dataset...
    if artist['BeginDate'] != "0":
        artists_with_year.append(artist)

print(len(artists_with_year))
```

```python
year_counts = {}  # empty DICTIONARY to store year counts

for artist in artists_with_year: # go through each artist in the filtered list
    year = artist['BeginDate'] #store the begin date for each artist
    if year in year_counts: #check to see if that begin date is in the year_counts dictionary
        year_counts[year] = year_counts[year]+1 # if true, add 1 to existing count 
    else:
        year_counts[year] = 1 #if that year isn't in the dictionary, create a key for the year and set value to 1

print(year_counts['1900'])
```

```python
sorted_years = sorted(year_counts, key=lambda year: year_counts[year], reverse=True)

print(sorted_years[:10])
```
